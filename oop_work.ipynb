```python
class Account:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def show_balance(self):
        return f'{self.name} has GHS {self.balance}'

    def deposit(self, amount):
        self.balance += amount
        return f"Deposited GHS: {amount}"

    def withdraw(self, amount):
        if amount > self.balance:
            return "Insufficient Funds"
        else:
            self.balance -= amount
            return "Withdrawal Successful"

# Child 1: Savings Account (Adds Interest)
class SavingsAccount(Account):
    def interest(self):
        self.balance *= 1.05
        return f"Interest added. New Balance GHS: {self.balance}"

# Child 2: Current Account (Allows Overdraft)
class CurrentAccount(Account):
    def withdraw(self, amount):
        # Allow withdrawal if amount is within balance + 500 overdraft
        if amount > (self.balance + 500):
            return "Overdraft limit exceeded!"
        else:
            self.balance -= amount
            return "Withdrawal Successful (Overdraft used)"

class Bank:
    # FIXED: Added 'name' here so it matches your code at the bottom
    def __init__(self, name):
        self.name = name
        self.accounts = []

    def add_account(self, account):
        self.accounts.append(account)

    def get_total_assets(self):
        total = 0
        for account in self.accounts:
            total += account.balance
        return total

# --- TESTING THE CODE ---

# --- FINAL BOSS TEST ---
print("\n--- The Bank Manager ---")
gcb = Bank("Ghana Commercial Bank")

# Create two accounts
ama_acc = SavingsAccount("Ama", 5000)
kojo_acc = CurrentAccount("Kojo", -200) # He owes money

# Add them to the bank
gcb.add_account(ama_acc)
gcb.add_account(kojo_acc)

# Calculate total (5000 + -200 = 4800)
print(f"Total Assets in {gcb.name}: GHS {gcb.get_total_assets()}")


# 1. Standard Account Test
print("\n--- Standard Account ---")
acc1 = Account('Jehosaphat', 1000)
print(acc1.show_balance())
print(acc1.deposit(500))
print(acc1.withdraw(2000)) # Fails (Insufficient funds)
print(acc1.show_balance())

# 2. Savings Account Test
print("\n--- Savings Account ---")
save1 = SavingsAccount('Kwame', 2000)
print(save1.interest())
print(save1.show_balance())

# 3. Current Account Test
print("\n--- Current Account ---")
business_acc = CurrentAccount("Business", 100)
# Try to withdraw 400 (Balance is 100, but limit is 600 total)
print(business_acc.withdraw(400))
print(business_acc.show_balance()) # Should be -300
```
