```python
import xml.etree.ElementTree as ET

# create the file structure
employee = ET.Element('employee')
details = ET.SubElement(employee, 'details')
first = ET.SubElement(details, 'firstname')
second = ET.SubElement(details, 'lastname')
third = ET.SubElement(details, 'age')
first.text = 'Shiv'
second.text = 'Mishra'
third.text = '23'

# create a new XML file with the results
mydata1 = ET.ElementTree(employee)
# myfile = open("items2.xml", "wb")
# myfile.write(mydata)
with open("new_sample.xml", "wb") as files:
    mydata1.write(files)
```

```python
import requests
import xml.etree.ElementTree as etree

# URL of the XML file
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%205/data/Sample-employee-XML-file.xml"

# Download the file
response = requests.get(url)
with open("Sample-employee-XML-file.xml", "wb") as f:
    f.write(response.content)

# Parse the XML file
tree = etree.parse("Sample-employee-XML-file.xml")
root = tree.getroot()

# Example: print all employee names
for emp in root.findall("employee"):
    name = emp.find("firstname").text
    print(name)
```
